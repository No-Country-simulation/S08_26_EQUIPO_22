# Ingeniería de Variables Térmicas y Preparación de Datos de Sensores

## Objetivo

La etapa de ingeniería de variables busca transformar los datos originales provenientes de sensores industriales en información útil para modelos de detección de anomalías y predicción de fallas.

Los datos de sensores contienen mediciones dinámicas del comportamiento del equipo a través del tiempo, por lo cual es necesario realizar procesos de:

- Conversión de variables temporales.
- Validación de calidad de datos.
- Creación de indicadores térmicos.
- Extracción de características relacionadas con degradación del equipo.

Estas nuevas variables permitirán que los modelos de Machine Learning identifiquen patrones asociados con condiciones anormales de operación.

---

# 1. Conversión de variables temporales

La columna `observation_timestamp` representa el instante en que fue registrada cada medición del equipo.

Esta variable inicialmente se encuentra almacenada como texto (`string`), por lo cual debe convertirse al formato fecha/hora de Python (`datetime`).

La conversión permite realizar análisis como:

- Evolución histórica de temperatura.
- Tendencias de degradación.
- Comparación entre periodos de operación.
- Ordenamiento cronológico de las mediciones.

Ejemplo:

Antes:


2026-04-09T15:14:38

Después:

2026-04-09 16:14:38

## Planteamiento del problema y comprensión de la condición de falla

Los equipos industriales no suelen presentar una falla súbita; en la mayoría de los casos, el deterioro aparece progresivamente mediante cambios medibles en sus variables operacionales. El análisis inicial del dataset revela que la condición del activo no puede interpretarse únicamente como un estado binario de funcionamiento o falla, ya que existen diferentes etapas de degradación antes de alcanzar una condición crítica. Variables como temperatura de rodamientos, vibración, comportamiento eléctrico, condiciones de lubricación, carga operacional y parámetros del proceso permiten observar la evolución del equipo desde una operación estable hasta una condición anómala que puede comprometer su disponibilidad.

El análisis de las etiquetas disponibles evidencia que `failure_mode_active` identifica el mecanismo de degradación presente en el equipo, mientras que `failure_mode_severity` representa el nivel de afectación alcanzado. Por su parte, `anomaly_label` permite diferenciar aquellos estados donde la condición operacional supera los límites normales y requiere atención. Esto indica que una falla industrial no debe entenderse únicamente como un evento final, sino como una transición progresiva donde aparecen señales tempranas asociadas a desgaste mecánico, desbalance, problemas eléctricos, cavitación u otras condiciones anormales.

Por lo tanto, el objetivo del modelo predictivo no será simplemente identificar si un equipo está fallado, sino aprender los patrones de comportamiento que anteceden una condición anómala, permitiendo generar alertas tempranas y apoyar decisiones de mantenimiento preventivo. La hipótesis inicial plantea que las variables capturadas por los sensores industriales contienen suficiente información para identificar cambios en el estado de salud del activo antes de que ocurra una falla crítica, reduciendo el riesgo de paradas no planificadas y mejorando la gestión del ciclo de vida del equipo.

# Propuesta de solución: Modelo de detección temprana de condiciones anómalas

A partir del análisis realizado, se plantea desarrollar un modelo de aprendizaje automático supervisado cuyo objetivo sea identificar condiciones anómalas en equipos industriales antes de que evolucionen hacia una falla crítica.

La solución propuesta utiliza las variables provenientes de los sensores industriales como fuente principal de información, integrando parámetros térmicos, vibracionales, eléctricos, hidráulicos y de lubricación. Estas variables representan el comportamiento físico del activo durante su operación y permiten capturar señales tempranas asociadas con procesos de degradación.

El modelo tendrá como variable objetivo:


- `0` representa una condición operacional controlada.
- `1` representa una condición anómala que requiere atención.

Definir variable objetivo

````python
# Variable objetivo del modelo

target = "anomaly_label"

sensor[target].value_counts()
````

Definir variables candidatas

````python
features = [

    # Variables térmicas
    "temp_bearing_drive_end_c",
    "temp_bearing_non_drive_end_c",
    "temp_motor_winding_c",
    "thermal_gradient_c_per_hr",
    "temp_delta_bearing_c",
    "thermal_load",

    # Vibración
    "vib_overall_rms_mm_s",
    "vib_peak_mm_s",
    "vib_crest_factor",
    "vib_kurtosis",
    "vib_bpfo_amplitude",
    "vib_bpfi_amplitude",

    # Variables hidráulicas
    "pressure_differential_bar",
    "pressure_lube_oil_bar",
    "flow_rate_m3_hr",
    "cavitation_index",

    # Variables eléctricas
    "current_imbalance_pct",
    "power_factor",
    "active_power_kw",
    "motor_load_pct",

    # Lubricación
    "oil_viscosity_cst",
    "oil_contamination_ntu",
    "hours_since_lubrication",

    # Operación
    "operating_speed_rpm",
    "efficiency_pct"
]
````
